```python
df = df.dropna().reset_index(drop=True)

print(df.shape)
```

```python
df = df.drop(columns=["CustomerID"])
```

```python
df["Churn"].value_counts()
```

```python
df["Churn"].value_counts(normalize=True) * 100
```

```python
numeric_cols = [
    "Age",
    "Tenure",
    "Usage Frequency",
    "Support Calls",
    "Payment Delay",
    "Total Spend",
    "Last Interaction"
]

df.groupby("Churn")[numeric_cols].mean().round(2)
```

```python
features = [
    "Support Calls",
    "Payment Delay",
    "Total Spend",
    "Last Interaction"
]

df[features + ["Churn"]].corr()["Churn"].sort_values(ascending=False)
```
